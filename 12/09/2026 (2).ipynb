```python
def add(a, b=10):          # b has a default value
    return a + b

print(add(5, 3))    # 8
print(add(5))        # 15 (uses default b=10)

def multiply_all(*args):        # variable-length positional args
    result = 1
    for num in args:
        result *= num
    return result

print(multiply_all(2, 3, 4))    # 24

def show_info(**kwargs):        # variable-length keyword args
    for key, value in kwargs.items():
        print(key, ":", value)

show_info(name="Harry", age=25)
```
